```python
from experiments.chess_perst.chess_engine import EngineConfig

engine_config = EngineConfig.model_validate(
    {
        "limit": {"time": 0.5},
        "engine_path": "stockfish",
        "threads": 3,
        "hash_mb": 64,
        "reward": "win_prob",
        "eval_mode": "root_multipv",
        "illegal_score": 0.0,
        "mate_score": 10000,
    }
)
```

```python
from experiments.chess_perst.chess_engine import MoveEvaluator, MoveResult

evaluator = MoveEvaluator.for_process(engine_config)
```

```python
from experiments.chess_perst.chess_engine import MoveEvaluator, MoveResult
from experiments.chess_perst.rewards import compute_reward
from chess import Board

fens = [
    "r4rk1/5ppp/pQ1b1q2/8/1P6/P4N1P/5PP1/3R1RK1 b - - 2 25",
    "r5k1/2n5/5r1p/p1pP1N2/PpP2p2/1P1Q4/3B1q1N/7K w - - 0 33",
    "R4bk1/5rP1/5pR1/8/2r5/p5P1/6K1/1q6 w - - 0 42",
    "2r4k/p5pp/1p2q3/2p3Q1/8/7P/PB3PP1/6K1 w - - 0 30",
    "2Q5/1p3kp1/3p2p1/3Pp3/2N1n3/P7/KQq5/8 b - - 1 33",
    "1rr5/pbRp3p/1p1nkpp1/4p3/1B2P1P1/3B1P1P/PP2K3/2R5 w - - 8 25",
    "Q7/4ppbk/6pp/5q2/P3n3/4PNBP/1r3PP1/3R2K1 b - - 0 31",
    "1r3rk1/pbp3bp/6p1/2qN1p2/1P2B3/5Q2/P1P2PPP/1R3RK1 b - - 0 16",
]

for i, fen in enumerate(fens):
    print(f"Position {i+1}: {fen}")
    legal_moves = [move.uci() for move in Board(fen).legal_moves]
    
    for move in legal_moves:
        result = await evaluator.score(fen, move)
        reward = compute_reward(result, engine_config)
        print(f"{move}: {result.score} {reward:.4f}")
```
